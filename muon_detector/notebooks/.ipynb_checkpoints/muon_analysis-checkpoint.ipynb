```python
cal005 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_0-5mus.txt')
cal02 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_2mus.txt')
cal03 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_3mus.txt')
cal04 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_4mus.txt')
cal05 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_5mus.txt')
cal06 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_6mus.txt')
cal07 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_7mus.txt')
cal08 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_8mus.txt')
cal09 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/calibration/cal_9mus.txt')

calibration_dfs = [cal005, cal02, cal03, cal04, cal05, cal06, cal07, cal08, cal09]
```

```python
#defining a function to find the channel in which the peak occurs, and the width on either side for errors
def get_channel_err(df):
    peak = np.max(df['0'])
    channel_no = np.where(df['0']==peak)[0][0]

    peak_area = df[df['0']>0]

    low_err = channel_no - peak_area.index[0]
    high_err = peak_area.index[-1] - channel_no

    return channel_no, low_err, high_err
    
```

```python
channels = []
cal_low_errs = []
cal_high_errs = []

for df in calibration_dfs:
    channel_no, low_err, high_err = get_channel_err(df)
    channels.append(channel_no)
    cal_low_errs.append(low_err)
    cal_high_errs.append(high_err)

times = [0.5, 2, 3, 4, 5, 6, 7, 8, 9]
```

```python
#fitting a calibration line to the data
def lin(x, a, b):
    return a*x + b

popt, pcov = curve_fit(lin, channels, times)
a_time, b_time = popt
a_err, b_err = np.sqrt(np.diag(pcov))
print(a_time, b_time)
```

```python
fig, ax = plt.subplots()

ax.scatter(channels, times)

cal_err = [cal_low_errs, cal_high_errs]
ax.errorbar(channels, times, xerr = cal_err, fmt='None', color='black')

plt.plot(channels, lin(np.array(channels), a_time, b_time), color='darkgray')

ax.set_xlabel('Channel number')
ax.set_ylabel(r'Time ($\mu$s)')
```

```python
def channel_time(channel_no):
    return a_time*channel_no + b_time

def time_channel(time):
    return (time - b_time)/a_time
```

```python
#defining a decay function to fit to the data, with b = lifetime, c = constant background term 
def decay(x, a, b, c):
    return a*np.exp(-x/b) + c
```

```python
data1 = pd.read_csv('C:/Users/dearb/Documents/UCD_labs/fourth_year_labs/muon_detector/data/thurs_15_0115.txt', header=None)
```

```python
channels = np.array(data1.index)
counts = data1[0].values

bin_lower_edges = []
bin_upper_edges = []
channel_sizes = []
bin_centres = []
bin_averages = []

exp=decay(channels, 60, 2.197, 2)

i = len(channels)-1
while i>0:
    exp_totals = exp[i]
    count_totals = counts[i]
    n=0
    while exp_totals<10:
        n+=1
        exp_totals+=exp[i-n]
        count_totals+=counts[i-n]
        
    bin_lower_edges.append(channels[i])
    bin_upper_edges.append(channels[i-n])
    channel_sizes.append(n+1)
    
    if n>0:
        bin_centres.append((channels[i] + channels[i-n])/2)
        bin_averages.append(count_totals/(n+1))
    else:
        bin_centres.append(channels[i])
        bin_averages.append(count_totals)
    i-=n+1
```
